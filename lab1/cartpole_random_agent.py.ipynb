```python
"""
Reinforcement Learning Lab 1
Name: Mahadevi E Malkhed
Roll No: BL.SC.P2CSE25009

Exploring Reinforcement Learning Environments using Gymnasium.
"""

import gymnasium as gym


def main():
    # Create the CartPole-v1 environment
    env = gym.make("CartPole-v1")

    print("Environment:", env.spec.id)
    print("Observation space:", env.observation_space)
    print("Action space:", env.action_space)

    # Start an episode
    observation, info = env.reset(seed=42)

    total_reward = 0.0
    steps = 0
    terminated = False
    truncated = False

    # Random agent: sample an action from the environment's action space
    while not (terminated or truncated):
        action = env.action_space.sample()
        observation, reward, terminated, truncated, info = env.step(action)

        total_reward += reward
        steps += 1

    print("Episode steps:", steps)
    print("Cumulative reward:", total_reward)

    env.close()


if __name__ == "__main__":
    main()
```
